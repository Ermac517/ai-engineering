### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

### Step 2: Call the API

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Next week I'm traveling to Spain. Can you give me some tips?"},
    ],
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### In theory... (doing it manually for illustration)

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Next week I'm traveling to Spain."},
    {"role": "assistant", "content": "That sounds exciting! Do you need any tips for your trip?"},
    {"role": "user", "content": "Where am I travelling to next week?"}
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
