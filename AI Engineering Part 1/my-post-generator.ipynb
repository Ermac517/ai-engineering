### Step 1: Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")

client = OpenAI(api_key=OPENAI_API_KEY)
```

### Step 2: Searching the web

```python
completion = client.chat.completions.create(
    model="gpt-4o-mini-search-preview",
    messages=[
        {"role": "user", "content": "What is today's date? Tell me the 3 top stories in AI in this past week, according to reliable news media, do not use simple news aggregators. Each story should be from a separate source and date. Include dates."}
    ],
    max_tokens=500
)
```

```python
search_results = completion.choices[0].message.content
display(Markdown(search_results))
```

### Step 3: Summarize and write social media post

```python
system_message = """
You are a helpful assistant who creates amazing LinkedIn posts about AI based on text that you are provided with. Make the post engaging and interesting.
Also make it concise and add a thought-provoking question at the end to encourage comments.
"""

prompt = f"""
The following text containts top 3 news stories about AI from the past week.
Now please write my LinkedIn post based on this information. Make it engaging and interesting.
{search_results}
"""

completion = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": prompt}
    ],
    max_tokens=500,
    temperature=0.7
)

chat_response = completion.choices[0].message.content
```

```python
print(chat_response)
```
