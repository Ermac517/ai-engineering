# Line 3: Specify which GPT model to use.

# Line 4: Start the messages list with conversation history.

# Line 5: System message sets the AI's behavior and context.

# Line 6: User message contains the actual prompt/question.

# Line 9: Extract the text content from the first response choice.

# Line 10: Display the response as formatted Markdown in the notebook.

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display
from pprint import pprint
import gradio as gr

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")

client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
messages = [
    {"role": "system", "content": "You are a helpful assistant."},
    {"role": "user", "content": "Next week I'm traveling to Spain."}
]

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

reply = response.choices[0].message.content
messages.append({"role": "assistant", "content": reply})

print(messages)

display(Markdown(reply))
```
