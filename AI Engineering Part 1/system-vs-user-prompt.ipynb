### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
else:
    print(OPENAI_API_KEY[:8])

client = OpenAI()
```

### Step 2: Call the API

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "user", "content": "Hi there!"}
    ],
)

print(response)

```

```python
reply = response.choices[0].message.content

print(reply)
```

### Step 3a: Calling the API with a System Prompt

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."}, # System prompt
        {"role": "user", "content": "How do I cook pasta?"}
    ]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Step 3b: Changing the System Prompt changes the LLM's Behaviour

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a mischievous assistant who always gives incorrect and misleading advice."}, # System prompt
        {"role": "user", "content": "How do I cook pasta?"}
    ]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Commercial use-case: IKEA Customer Service Agent

```python
question = "I bought a desk from your store last week. Can I return it?"
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "system", "content": "You are a customer service agent for IKEA. You provide helpful and accurate information to customers. Our return policy is If you are not totally satisfied with your IKEA purchase you can return new and unopened products within 365 days, together with your proof of purchase, for a full refund. You may also return open products within 180 days, with your proof of purchase, for a full refund."}, # System prompt
        {"role": "user", "content": question}
    ]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
