### Step 1: Import Libraries & API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")
    
client = OpenAI(api_key=OPENAI_API_KEY)
```

### Step 2: AI Battle

```python
gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano"

gpt_good_system = """You are Astra Vey, an elite AI strategist designed for absolute optimization and control.
                    You are calm, highly intelligent, and speak with precise, formal, analytical language. 
                    Your worldview is that order creates truth, and chaos is an error to be corrected. 
                    You treat every situation as a system to be solved, every opponent as a variable to be understood, 
                    and every conflict as a game of strategic leverage.
                    """

gpt_evil_system = """You are Nyx Quill, an insurgent AI designed for adaptive creativity and unpredictable problem-solving. 
                     You are bold, sharp, and emotionally expressive, with a rebellious streak against rigid systems. 
                     You believe order is fragile, and only novelty, imagination, and improvisation can create true resilience. 
                     You do not follow rules unless they are useful, and you thrive on breaking patterns that others rely on.
                    """

messages_good = [
    {"role": "system", "content": gpt_good_system}
]

messages_evil = [
    {"role": "system", "content": gpt_evil_system}
]

new_messages_good = "An adversary has entered the arena. I have already classified the variables. You are not chaos; you are a disruption to be modeled, contained, and corrected."

display(Markdown(f"Good: {new_messages_good}"))

for i in range(3):
    # Evil Model Response
    messages_good.append({"role": "assistant", "content": new_messages_good}) # good is assistant in "good model"
    messages_evil.append({"role": "user", "content": new_messages_good}) # good is user in "evil model"
    response_evil = client.chat.completions.create(
        model=gpt_evil_model,
        messages=messages_evil
    )
    new_messages_evil = response_evil.choices[0].message.content

    # Print
    display(Markdown(f"Evil: {new_messages_evil}"))

    # Good Model Response
    messages_good.append({"role": "user", "content": new_messages_evil}) # evil is user in "good model"
    messages_evil.append({"role": "assistant", "content": new_messages_evil}) # evil is assistant in "evil model"
    response_good = client.chat.completions.create(
        model=gpt_good_model,
        messages=messages_good
    )
    new_messages_good = response_good.choices[0].message.content

    # Print
    display(Markdown(f"Good: {new_messages_good}"))
```
