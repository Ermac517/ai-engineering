### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")

client = OpenAI(api_key=OPENAI_API_KEY)
```

### 2. Response API: basic call

```python
response1 = client.responses.create(
    model="gpt-4.1-mini",
    instructions="You are a helpful assistant.",
    input="Next week I'm traveling to Spain. Can you give me some tips?"
)

reply = response1.output_text
display(Markdown(reply))
```

### Step 3: Responses API conversation history

```python
response2 = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id=response1.id,
    instructions="You are a helpful assistant.",
    input="Where am I travelling next week?"
)

reply = response2.output_text
display(Markdown(reply))
```

```python
response3 = client.responses.create(
    model="gpt-4.1-mini",
    previous_response_id=response2.id,
    instructions="You are a helpful assistant.",
    input="What is the capital?"
)

reply = response3.output_text
display(Markdown(reply))
```

### Step 4: Built-In Tools

```python
tone = "dramatic and engaging"

instructions = f"""
You are a helpful assistant who creates amazing LinkedIn posts about AI based on text that you are provided with. Make the post engaging and interesting.
Also make it concise and add a thought-provoking question at the end to encourage comments.
Use the following tone:
{tone}
"""

prompt = f"""
Write about the 4 top stories in AI this past week, according to reliable news media. Include date and source for each story.
"""

resp = client.responses.create(
    model="gpt-6-luna",
    instructions=instructions,
    input=prompt,
    tools=[{"type": "web_search"}]
)

display(Markdown(resp.output_text))
```
