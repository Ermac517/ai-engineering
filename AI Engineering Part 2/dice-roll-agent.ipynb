```python
import random
import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing.")

client = OpenAI()

def roll_dice():
    return random.randint(1, 6)
```

```python
target = 6
max_iterations = 20
```

```python
history = ""
iteration = 0

while iteration < max_iterations:
    iteration += 1

    instructions = f"""
        You are playing a dice rolling game. You need to roll a {target} on the dice.
        You have made the following rolls so far: {history if history else "no rolls yet"}
        Based on what you have rolled so far, decice your next action:
        Option A: If you have achieved the target, response with exactly: DONE: <message describing how you achieved the target>
        Option B: If you have not yet achieved the target, respond with exactly: ROLL THE DICE
    """

    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[
            {"role": "user", "content": instructions}
        ]
    )
    
    result = response.choices[0].message.content

    # Process the result to see if the agent has decided whether to roll again or not.
    if "DONE" in result:
        print(result)
        break
    elif "ROLL THE DICE" in result:
        roll = roll_dice()
        print(f"Rolled a {roll}\n")
        history += f"Roll {iteration}: {roll}\n"
    else:
        print("Unexpected response:", result)
        break

print(f"Game over after {iteration - 1} iterations. Final history:\n{history}")
```
